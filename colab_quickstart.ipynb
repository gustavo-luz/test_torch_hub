# SSL-HAR model zoo, quickstart on Colab

Load any of the 36 pretrained HAR models from [Zenodo](https://zenodo.org/records/19301058) with one call, run inference, and check the accuracy against the paper.

Nothing is cloned by hand: `torch.hub` fetches `hubconf.py` from GitHub, the architectures come from [Minerva](https://github.com/discovery-unicamp/Minerva), and the weights come from Zenodo.

**Runtime**: CPU is enough. Sections 1 to 3 need no data. Section 4 downloads the DAGHAR benchmark (191 MB) to reproduce a published accuracy.

Paper: *Benchmarking Encoders and Self-Supervised Learning for Smartphone-Based Human Activity Recognition*, IEEE Access 2026.

```python
import torch

REPO = "gustavo-luz/test_torch_hub"   # moving to H-IAAC/benchmarking-encoders-ssl-har

print("torch  :", torch.__version__)
print("device :", "cuda" if torch.cuda.is_available() else "cpu")
```

## 2. What is in the zoo

36 combinations of SSL technique and encoder, each trained on one of six DAGHAR datasets, each with a `pretrained` and a `finetuned` checkpoint.

Keys read `<ssl>_<encoder>_<dataset>`, for example `lfr_ts2vec_ms`.

```python
catalog = torch.hub.load(REPO, "list_models", trust_repo=True)

print(f"{len(catalog)} models\n")
print(f"{'key':<28} {'dataset':<17} {'paper':>6}")
for key, info in sorted(catalog.items(), key=lambda kv: -(kv[1]['accuracy'] or 0)):
    print(f"{key:<28} {info['dataset']:<17} {info['accuracy']:>5}%")
```

## 3. Load a model and run it

Three shapes are available for every key:

| Call | What you get |
|---|---|
| `torch.hub.load(REPO, key)` | architecture only, random weights |
| `... role="pretrained"` | the SSL backbone, no labels used in training |
| `... role="finetuned", head=True` | the full 6-class classifier |

```python
MODEL = "lfr_ts2vec_ms"   # LFR + TS2Vec on MotionSense, best result in the paper

clf = torch.hub.load(REPO, MODEL, role="finetuned", head=True, trust_repo=True)
clf.eval()

# one window: 6 IMU channels (accel xyz, gyro xyz), 60 samples
window = torch.randn(1, 6, 60)
with torch.no_grad():
    logits = clf(window)

ACTIVITIES = {0: "sit", 1: "stand", 2: "walk", 3: "stair up", 4: "stair down", 5: "run"}
print("logits :", tuple(logits.shape))
print("predicted on random noise:", ACTIVITIES[int(logits.argmax())])
```

```python
# the SSL backbone on its own, to use as a feature extractor
backbone = torch.hub.load(REPO, MODEL, role="pretrained", trust_repo=True)
backbone.eval()

with torch.no_grad():
    features = backbone(torch.randn(4, 6, 60))
print("features:", tuple(features.shape))
print(f"parameters: {sum(p.numel() for p in backbone.parameters()):,}")
```

## 4. Reproduce a published accuracy

Downloads the DAGHAR standardized view (191 MB) and evaluates the model on the test split of its own dataset.

Only `standardized_view.zip` is needed. The record also holds `baseline_view.zip`, which uses 150-sample windows and does not fit these models.

```python
from pathlib import Path

DATA = Path("daghar/standardized_view")

if not DATA.is_dir():
    !wget -q --show-progress -O standardized_view.zip \
        "https://zenodo.org/records/13987073/files/standardized_view.zip?download=1"
    !unzip -q -o standardized_view.zip -d daghar/
    !rm standardized_view.zip

print(sorted(p.name for p in DATA.iterdir() if p.is_dir()))
```

```python
import pandas as pd

FOLDER = {
    "ms": "MotionSense", "uci": "UCI", "kh": "KuHar",
    "rwthigh": "RealWorld_thigh", "rwwaist": "RealWorld_waist", "wisdm": "WISDM",
}
CHANNELS = ["accel-x", "accel-y", "accel-z", "gyro-x", "gyro-y", "gyro-z"]


def test_split(dataset):
    """Test CSV as tensors: each row is one window of 6 channels by 60 samples."""
    df = pd.read_csv(DATA / FOLDER[dataset] / "test.csv")
    cols = [c for ch in CHANNELS for c in df.columns if c.startswith(ch + "-")]
    x = torch.tensor(df[cols].to_numpy("float32")).reshape(-1, 6, 60)
    y = torch.tensor(df["standard activity code"].to_numpy("int64"))
    return x, y


dataset = catalog[MODEL]["dataset"]
x, y = test_split(MODEL.split("_")[-1])
print(f"{dataset}: {len(y)} windows, x={tuple(x.shape)}")
```

```python
correct = 0
with torch.no_grad():
    for i in range(0, len(x), 256):
        correct += (clf(x[i : i + 256]).argmax(1) == y[i : i + 256]).sum().item()

measured = round(100 * correct / len(y), 1)
paper = catalog[MODEL]["accuracy"]

print(f"model    : {MODEL}")
print(f"measured : {measured}%")
print(f"paper    : {paper}%")
print(f"delta    : {measured - paper:+.1f} pp")
```

Accuracy is compared at one decimal, the precision Table III prints, so 0.1 is a single step of the last digit. Across the whole catalog all 36 models land within 0.1 pp of their published number and 33 match it exactly.

Change `MODEL` in section 3 and rerun from there to check another one. To evaluate a model on a dataset it was not trained on, pass a different key to `test_split`.

## 5. Fine-tune from a pretrained backbone

The `pretrained` checkpoints are the SSL backbones, trained without labels. Attach a head and train it on your own data.

The sketch below is the shape of it, not a full training loop: a real run wants a validation split, early stopping and the learning rate from the paper.

```python
from minerva.models.nets.base import SimpleSupervisedModel
from minerva.models.nets.mlp import MLP
from minerva.models.adapters import MaxPoolingTransposingSqueezingAdapter

# TS2Vec returns (batch, time, features) outside TFC, so it needs the adapter
model = SimpleSupervisedModel(
    backbone=torch.hub.load(REPO, MODEL, role="pretrained", trust_repo=True),
    fc=MLP([320, 128, 6]),
    loss_fn=torch.nn.CrossEntropyLoss(),
    adapter=MaxPoolingTransposingSqueezingAdapter(kernel_size=60),
    flatten=False,
)

# freeze the backbone for linear probing, or leave it trainable for full finetuning
for p in model.backbone.parameters():
    p.requires_grad = False

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"trainable {trainable:,} of {total:,} parameters")
print("model is a LightningModule, use it with a lightning Trainer")
```

## Notes

- `torch.hub` caches the repository under `~/.cache/torch/hub`, so a new commit needs `force_reload=True`.
- Checkpoints are cached under `~/.cache/torch/hub/checkpoints`, between 1.6 MB and 84 MB each.
- `dependencies` in `hubconf.py` only checks that a module can be imported, it never installs anything, which is why section 1 comes first.
- The head sizes are read from the checkpoint itself, so `head=True` works without a lookup table.
